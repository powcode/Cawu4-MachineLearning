# Menggunakan Algoritma OLS

```python
#Define X
X = np.array([
    [50, 2],
    [60, 2],
    [70, 3],
    [80, 3],
    [90, 4]
])
```

```python
#Define y
y = np.array([250, 300, 350, 400, 450])
```

```python
# Tambahkan bias di kolom awal variabel X
X = np.c_[np.ones(X.shape[0]), X]
```

```python
# Gunakan perhitungan w = (xTx)^1XT y
w = np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
w
```

# Menggunakan Gradient Descent

```python
# Jumlah sampel dan fitur
m, n = X.shape # n jumlah kolom, m jumlah baris
```

```python
X.shape
```

```python
# Initialize weights and bias

w = np.zeros(n)
```

```python
#Choose learning rate and iterations
alpha = 0.0003
N_iter = 10000
```

```python
for i in range(N_iter):

    # Compute predicted values
    y_pred = np.dot(X, w)

    # Compute loss (MSE)
    J = (1 / (2 * m)) * np.sum((y_pred - y) ** 2)

    # Compute gradient with respect to weights
    dw = (1 / m) * np.dot(X.T, (y_pred - y))

    # Update weights
    w = w - alpha * dw

    # Display loss every 1000 iterations
    if (i + 1) % 100 == 0:
        print(f"Iteration {i+1}, Loss: {J:.4f}")
```
